##### Convert crash_date and crash_time to datetime

```python
df["crash_date"] = pd.to_datetime(df["crash_date"])
```

```python
df["crash_time"] = pd.to_datetime(df["crash_time"].astype(str), format="%H:%M:%S")
```

```python
df.dtypes
```

###### Extract Month and Day of the week from crash_date

```python
df["month_of_crash"] = df["crash_date"].dt.strftime("%b")
```

```python
df["day_of_crash"] = df["crash_date"].dt.day_name()
```

```python
df["month_of_crash"].head()
```

```python
df["day_of_crash"].head()
```

###### Extract Hour from crash_time

```python
df["hour_of_crash"] = df["crash_time"].dt.hour
```

```python
df["hour_of_crash"].head()
```
